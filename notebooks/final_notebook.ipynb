### Data Cleaning

````python
## Checking for Duplicates
# Check for duplicate rows
duplicate_count = df_original.duplicated().sum()
print(f"Number of duplicate rows: {duplicate_count}")

# Remove duplicates if any exist
if duplicate_count > 0:
    df = df_original.drop_duplicates()
    print(f"Data shape after removing duplicates: {df.shape}")
else:
    df = df_original.copy()
    print("No duplicates found in the dataset")
````

````python
## Data Type Validation
# Check data types of each column
print("\nData Types of Columns:")
print(df.dtypes)

# Ensure correct data types for each column
numeric_columns = ['BMI', 'HeightInMeters', 'WeightInKilograms', 
                  'PhysicalHealthDays', 'MentalHealthDays']
categorical_columns = [col for col in df.columns if col not in numeric_columns]

# Verify numeric columns contain only numbers
for col in numeric_columns:
    invalid_entries = df[~df[col].apply(lambda x: isinstance(x, (int, float)))][col]
    if len(invalid_entries) > 0:
        print(f"\nInvalid entries in {col}:")
        print(invalid_entries)
````

````python
## Value Range Validation
# Check value ranges for numeric columns
print("\nValue Ranges for Numeric Columns:")
for col in numeric_columns:
    print(f"\n{col}:")
    print(f"Min: {df[col].min()}")
    print(f"Max: {df[col].max()}")
    print(f"Mean: {df[col].mean():.2f}")
    
    # Identify potential outliers using IQR method
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)][col]
    if len(outliers) > 0:
        print(f"Number of potential outliers: {len(outliers)}")
````

````python
## Category Validation
# Check unique values in categorical columns
print("\nUnique Values in Categorical Columns:")
for col in categorical_columns:
    print(f"\n{col}:")
    print(df[col].value_counts())
    
    # Check for unexpected categories
    if col in ['Sex', 'GeneralHealth', 'HadHeartAttack']:
        expected_categories = {
            'Sex': ['Male', 'Female'],
            'GeneralHealth': ['Excellent', 'Very Good', 'Good', 'Fair', 'Poor'],
            'HadHeartAttack': ['Yes', 'No']
        }
        unexpected = set(df[col].unique()) - set(expected_categories[col])
        if unexpected:
            print(f"Unexpected categories found: {unexpected}")
````

````python
## 5. Data Consistency Checks
# Check for logical consistency
# Example: BMI consistency with height and weight
df['Calculated_BMI'] = df['WeightInKilograms'] / (df['HeightInMeters'] ** 2)
bmi_difference = abs(df['BMI'] - df['Calculated_BMI'])
inconsistent_bmi = df[bmi_difference > 1]  # threshold of 1 unit difference

if len(inconsistent_bmi) > 0:
    print("\nInconsistent BMI calculations found:")
    print(f"Number of inconsistencies: {len(inconsistent_bmi)}")


df = df.drop('Calculated_BMI', axis=1)
print("Calculated_BMI column has been dropped")
````

````python
# Calculate and Display Outlier Statistics
def calculate_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    return len(outliers), lower_bound, upper_bound

# Calculate outliers for each numerical column
numerical_cols = df_original.select_dtypes(include=['int64', 'float64']).columns
print("\nOutlier Analysis Summary:")
print("-" * 50)
for col in numerical_cols:
    n_outliers, lower, upper = calculate_outliers(df, col)
    print(f"\n{col}:")
    print(f"Number of outliers: {n_outliers}")
    print(f"Lower bound: {lower:.2f}")
    print(f"Upper bound: {upper:.2f}")
    print(f"Percentage of outliers: {(n_outliers/len(df))*100:.2f}%")
````

````python
# Function untuk menghapus outliers menggunakan metode IQR
def remove_outliers_iqr(df, columns):
    df_clean = df.copy()
    for column in columns:
        Q1 = df_clean[column].quantile(0.25)
        Q3 = df_clean[column].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        df_clean = df_clean[(df_clean[column] >= lower_bound) & (df_clean[column] <= upper_bound)]
    return df_clean
````

````python
# Define function to calculate Cramér's V
def cramers_v(x, y):
    confusion_matrix = pd.crosstab(x, y)
    chi2 = chi2_contingency(confusion_matrix)[0]
    n = confusion_matrix.sum().sum()
    min_dim = min(confusion_matrix.shape) - 1
    return np.sqrt(chi2 / (n * min_dim))

# Select categorical variables related to health conditions
categorical_health_vars = ['HighBloodPressure', 'HighCholesterol', 'HadDiabetes', 
                         'HadStroke', 'HadKidneyDisease', 'HadAsthma', 
                         'HadAngina', 'GeneralHealth', 'HadArthritis',
                         'HadCOPD','DifficultyWalking', 'HadHeartAttack']

# Create correlation matrix
n_vars = len(categorical_health_vars)
cramer_matrix = np.zeros((n_vars, n_vars))

for i in range(n_vars):
    for j in range(n_vars):
        cramer_matrix[i,j] = cramers_v(df_clean[categorical_health_vars[i]], 
                                     df_clean[categorical_health_vars[j]])

# Create heatmap
plt.figure(figsize=(12, 8))

# Create the heatmap with a more appealing color palette
sns.heatmap(pd.DataFrame(cramer_matrix, 
                        index=categorical_health_vars,
                        columns=categorical_health_vars),
            annot=True, cmap='YlGnBu', vmin=0, vmax=1, 
            cbar_kws={'label': 'Cramér\'s V Value'}, linewidths=1, linecolor='black')

# Add a title and labels with adjusted font sizes
plt.title('Cramér\'s V Correlation between Health Conditions', fontsize=18, pad=20)
plt.xlabel('Health Conditions', fontsize=14)
plt.ylabel('Health Conditions', fontsize=14)

# Show the plot
plt.tight_layout()
plt.show()
````

````python
# Select numeric columns
numeric_cols = ['BMI', 'PhysicalHealthDays', 'MentalHealthDays', 'HeightInMeters', 'WeightInKilograms']

# Create empty matrices for correlation coefficients and p-values
n = len(numeric_cols)
corr_matrix = np.zeros((n, n))
p_matrix = np.zeros((n, n))

# Calculate correlation coefficients and p-values
for i in range(n):
    for j in range(n):
        # Use Pearson for correlations between continuous variables
        corr, p_value = stats.pearsonr(df_clean[numeric_cols[i]], df_clean[numeric_cols[j]])
        corr_matrix[i, j] = corr
        p_matrix[i, j] = p_value

# Create a DataFrame for better visualization
corr_df = pd.DataFrame(corr_matrix, index=numeric_cols, columns=numeric_cols)

# Create mask for p-values > 0.05 (not significant)
mask = p_matrix > 0.05

# Create heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(corr_df, 
            annot=True, 
            cmap='coolwarm', 
            mask=mask,
            center=0,
            fmt='.3f',
            square=True,
            linewidths=0.5)

plt.title('Correlation Heatmap\n(Only showing correlations with p < 0.05)', pad=20)
plt.tight_layout()
plt.show()

# Print detailed correlation analysis for each numeric variable vs HadHeartAttack
print("\nDetailed Analysis with Heart Attack:")
for col in numeric_cols:
    heart_attack_yes = df_clean[df_clean['HadHeartAttack'] == 'Yes'][col]
    heart_attack_no = df_clean[df_clean['HadHeartAttack'] == 'No'][col]
    stat, p_value = stats.ttest_ind(heart_attack_yes, heart_attack_no)
    if p_value < 0.05:
        print(f"\n{col}:")
        print(f"Mean for Heart Attack = Yes: {heart_attack_yes.mean():.3f}")
        print(f"Mean for Heart Attack = No: {heart_attack_no.mean():.3f}")
        print(f"T-statistic: {stat:.3f}")
        print(f"P-value: {p_value:.3e}")
````

````python
# Select numeric columns
numeric_cols = ['BMI', 'PhysicalHealthDays', 'MentalHealthDays', 'HeightInMeters', 'WeightInKilograms']

# Convert HadHeartAttack to numeric (0/1)
target = (df_clean['HadHeartAttack'] == 'Yes').astype(int)

# Calculate Point-Biserial correlation and p-values
correlations = []
p_values = []
for col in numeric_cols:
    corr, p_val = stats.pointbiserialr(target, df_clean[col])
    correlations.append(abs(corr))  # Use absolute correlation values
    p_values.append(p_val)

# Create correlation matrix visualization
plt.figure(figsize=(12, 8))

# Create bars starting from 0
bars = plt.bar(numeric_cols, correlations, color='red')

# Set y-axis to start from 0
plt.ylim(0, max(correlations) * 1.4)  # Add 40% padding at top

# Add correlation values on top of bars
for i, (bar, corr, p_val) in enumerate(zip(bars, correlations, p_values)):
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, 
             height + (max(correlations) * 0.05),
             f'r={corr:.3f}\np={p_val:.2e}',
             ha='center', va='bottom',
             fontsize=10)

plt.title('Absolute Point-Biserial Correlation with Heart Attack', fontsize=12, pad=20)
plt.xlabel('Numeric Variables', fontsize=10)
plt.ylabel('Absolute Correlation Coefficient', fontsize=10)
plt.xticks(rotation=45, ha='right')
plt.grid(True, axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()

# Print detailed analysis
print("\nDetailed Correlation Analysis:")
for col, corr, p_val in zip(numeric_cols, correlations, p_values):
    print(f"\n{col}:")
    print(f"Absolute correlation coefficient: {corr:.3f}")
    print(f"P-value: {p_val:.2e}")
    
    # Calculate means for interpretation
    mean_yes = df_clean[df_clean['HadHeartAttack'] == 'Yes'][col].mean()
    mean_no = df_clean[df_clean['HadHeartAttack'] == 'No'][col].mean()
    print(f"Mean for Heart Attack = Yes: {mean_yes:.3f}")
    print(f"Mean for Heart Attack = No: {mean_no:.3f}")
````

````python
import logging
from pathlib import Path

class HeartAttackPredictor:
    """A class for predicting heart attack risk using ensemble models."""
    
    def __init__(self, model_path='../models/'):
        """Initialize predictor with path to models."""
        self.model_path = Path(model_path)
        self._setup_logging()
        self._load_components()
        
        # Update features to match your dataset
        self.features = ['Sex', 'GeneralHealth', 'HighCholesterol', 
                        'HighBloodPressure', 'HadAngina', 'HadStroke', 
                        'HadCOPD', 'HadKidneyDisease', 'DifficultyWalking', 
                        'SmokerStatus', 'AgeCategory', 'HadDiabetes']

    def _setup_logging(self):
        """Configure logging."""
        logging.basicConfig(
            level=logging.INFO,
            format='%(asctime)s - %(levelname)s - %(message)s'
        )
        self.logger = logging.getLogger(__name__)

    def _load_components(self):
        """Load models and mappings."""
        try:
            # Load models
            self.models = {
                'xgb': self._load_model('best_xgb.pkl'),
                'rf': self._load_model('best_rf.pkl'),
                'lr': self._load_model('best_lr.pkl')
            }
            
            # Define mappings
            self._define_mappings()
            
            self.logger.info("Successfully loaded all components")
        except Exception as e:
            self.logger.error(f"Error loading components: {str(e)}")
            raise

    def _load_model(self, filename):
        """Load a single model from file."""
        try:
            with open(self.model_path / filename, 'rb') as f:
                return pickle.load(f)
        except Exception as e:
            self.logger.error(f"Error loading {filename}: {str(e)}")
            raise

    def _define_mappings(self):
        """Define encoding mappings."""
        # Update mappings based on your dataset
        self.ordinal_mapping = {
            'GeneralHealth': {
                "Excellent": 5, "Very good": 4, "Good": 3, 
                "Fair": 2, "Poor": 1
            },
            'SmokerStatus': {
                "Current smoker - now smokes every day": 4,
                "Current smoker - now smokes some days": 3,
                "Former smoker": 2, 
                "Never smoked": 1
            },
            'AgeCategory': {
                "Age 18 to 24": 1, "Age 25 to 29": 2,
                "Age 30 to 34": 3, "Age 35 to 39": 4,
                "Age 40 to 44": 5, "Age 45 to 49": 6,
                "Age 50 to 54": 7, "Age 55 to 59": 8,
                "Age 60 to 64": 9, "Age 65 to 69": 10,
                "Age 70 to 74": 11, "Age 75 to 79": 12,
                "Age 80 or older": 13
            },
            'HadDiabetes': {
                'No': 1, 
                'No, pre-diabetes or borderline diabetes': 2,
                'Yes, but only during pregnancy (female)': 3, 
                'Yes': 4
            },
            'HighBloodPressure': {
                'No': 1,
                'No, pre-hypertension or borderline high blood pressure': 2,
                'Yes, but female told only during pregnancy': 3,
                'Yes': 4
            }
        }
        self.binary_mapping = {'No': 0, 'Yes': 1}

    def _validate_input(self, df):
        """Validate input data format and required columns."""
        missing_cols = set(self.features) - set(df.columns)
        if missing_cols:
            raise ValueError(f"Missing required columns: {missing_cols}")

    def _preprocess_input(self, data):
        """Preprocess input data for prediction."""
        df = data.copy()
        
        # Apply binary encoding
        binary_cols = ['HadAngina', 'HadStroke', 'HighCholesterol',
                      'HadCOPD', 'DifficultyWalking', 'HadKidneyDisease']
        for col in binary_cols:
            df[col] = df[col].map(self.binary_mapping)
        
        # Apply ordinal encoding
        for col, mapping in self.ordinal_mapping.items():
            df[col] = df[col].map(mapping)
        
        # Encode Sex
        df['Sex'] = df['Sex'].map({'Male': 1, 'Female': 0})
        
        # Ensure correct column order
        df = df[self.features]
        
        return df

    def predict(self, data, threshold=0.5):
        """Make ensemble prediction."""
        try:
            self._validate_input(data)
            processed_data = self._preprocess_input(data)
            
            # Get predictions from each model
            predictions = {}
            for name, model in self.models.items():
                pred_proba = model.predict_proba(processed_data)[:, 1]
                predictions[name] = pred_proba[0]
                self.logger.info(f"{name} prediction: {pred_proba[0]:.3f}")

            # Calculate ensemble prediction
            final_proba = np.mean(list(predictions.values()))
            final_pred = int(final_proba > threshold)
            
            # Get risk analysis
            risk_factors = self._identify_risk_factors(data)
            risk_level = self._get_risk_level(final_proba)
            
            return {
                'prediction': bool(final_pred),
                'probability': float(final_proba),
                'risk_factors': risk_factors,
                'risk_level': risk_level,
                'model_predictions': predictions
            }
            
        except Exception as e:
            self.logger.error(f"Prediction error: {str(e)}")
            raise

    def _identify_risk_factors(self, data):
        """Identify main risk factors from input data."""
        risk_factors = []
        
        # Check age risk
        if data['AgeCategory'].iloc[0] in ['Age 65 to 69', 'Age 70 to 74', 
                                           'Age 75 to 79', 'Age 80 or older']:
            risk_factors.append('Advanced Age')
        
        # Check health conditions
        conditions = {
            'HighBloodPressure': 'High Blood Pressure',
            'HighCholesterol': 'High Cholesterol',
            'HadDiabetes': 'Diabetes',
            'HadAngina': 'Angina',
            'HadStroke': 'Previous Stroke',
            'HadCOPD': 'COPD',
            'HadKidneyDisease': 'Kidney Disease'
        }
        
        for col, condition in conditions.items():
            if data[col].iloc[0] == 'Yes':
                risk_factors.append(condition)
        
        # Check other risk factors
        if data['DifficultyWalking'].iloc[0] == 'Yes':
            risk_factors.append('Difficulty Walking')
            
        if data['GeneralHealth'].iloc[0] in ['Poor', 'Fair']:
            risk_factors.append('Poor General Health')
            
        if data['SmokerStatus'].iloc[0].startswith('Current smoker'):
            risk_factors.append('Current Smoker')
            
        return risk_factors

    def _get_risk_level(self, probability):
        """Convert probability to risk level."""
        if probability < 0.2:
            return 'Low Risk'
        elif probability < 0.4:
            return 'Moderate Risk'
        elif probability < 0.6:
            return 'High Risk'
        else:
            return 'Very High Risk'

def demo_predictor():
    """Demonstrate predictor usage with correct features."""
    # Create sample data with correct features
    sample_data = pd.DataFrame({
        'Sex': ['Male'],
        'GeneralHealth': ['Fair'],
        'HighCholesterol': ['Yes'],
        'HadAngina': ['No'],
        'HadStroke': ['No'],
        'HadCOPD': ['No'],
        'HadKidneyDisease': ['No'],
        'DifficultyWalking': ['No'],
        'SmokerStatus': ['Current smoker - now smokes every day'],
        'AgeCategory': ['Age 65 to 69'],
        'HadDiabetes': ['Yes'],
        'HighBloodPressure': ['Yes']
    })

    try:
        predictor = HeartAttackPredictor()
        result = predictor.predict(sample_data)
        
        print("\n🫀 Heart Attack Risk Assessment")
        print("=" * 50)
        print(f"📊 Risk Level: {result['risk_level']}")
        print(f"📈 Risk Probability: {result['probability']:.2%}")
        print("\n🚩 Identified Risk Factors:")
        for factor in result['risk_factors']:
            print(f"• {factor}")
        print("\n🤖 Model Predictions:")
        for model, prob in result['model_predictions'].items():
            print(f"• {model}: {prob:.2%}")
            
    except Exception as e:
        print(f"❌ Error during prediction: {str(e)}")

if __name__ == "__main__":
    demo_predictor()
````
